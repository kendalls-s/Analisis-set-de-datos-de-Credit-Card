
# 3.1 Preparacion de los datos y por que usamos log1p

 Antes de agrupar clientes tenemos que dejar los datos "parejos". Recordemos
 lo que vimos en el EDA: las variables de dinero (`BALANCE`, `PURCHASES`,
 `CASH_ADVANCE`, `PAYMENTS`...) estaban todas cargadas a la izquierda con una
 cola muy larga a la derecha. La mayoria de la gente maneja montos pequenos y
 unos poquitos clientes mueven cifras enormes.

 El problema es que **K-Means agrupa por distancia**: mide que tan lejos esta
 un cliente de otro. Si una variable llega a valores gigantes, esa sola
 variable "grita" mas fuerte que las demas y termina mandando en los grupos.
 Es como comparar personas mezclando su edad (0 a 100) con su salario
 (0 a millones): el salario aplasta a la edad.

 La transformacion `log1p` (que es simplemente $\log(1+x)$) **comprime la cola**:
 acerca a los clientes gigantes al resto sin borrarlos, y deja la distribucion
 con una forma mucho mas parecida a una campana. Usamos `1+x` para poder incluir
 el valor 0 sin que reviente el logaritmo. Solo se la aplicamos a las variables
 de dinero; las frecuencias (que ya van de 0 a 1), los conteos y `TENURE` se
 quedan igual.

```python
# %% Carga del dataset limpio y definicion de las variables de dinero
datos = pd.read_csv('CC_GENERAL_clean.csv')

# Variables monetarias: tienen cola larga a la derecha y no tienen tope.
monetarias = ['BALANCE', 'PURCHASES', 'ONEOFF_PURCHASES', 'INSTALLMENTS_PURCHASES',
              'CASH_ADVANCE', 'CREDIT_LIMIT', 'PAYMENTS', 'MINIMUM_PAYMENTS']

print('Dimensiones del dataset:', datos.shape)
print('Variables a las que aplicaremos log1p:', len(monetarias))


# %% Version SIN log1p  y  version CON log1p
# Version A (sin log1p): los datos tal cual.
datos_sin = datos.copy()

# Version B (con log1p): solo transformamos las variables de dinero.
datos_con = datos.copy()
datos_con[monetarias] = np.log1p(datos_con[monetarias])

print('Ejemplo BALANCE  -> sin log1p:', round(datos_sin['BALANCE'].max(), 1),
      '| con log1p:', round(datos_con['BALANCE'].max(), 2))
```

# Lectura comparativa visual
 La diferencia salta a la vista. En la **columna izquierda (sin log1p)** casi
todas las barras se amontonan pegadas al cero y la cola se estira tanto que el
grafico queda "vacio" a la derecha: hay clientes tan grandes que estiran toda
 la escala. Asi, cuando midamos distancias, esos pocos clientes gigantes van a
 decidir los grupos casi ellos solos.

 En la **columna derecha (con log1p)** la misma variable se ve mucho mas
 repartida, con una forma parecida a una campana. Los clientes grandes siguen
 estando (no los borramos), pero ya no aplastan al resto. Esta es la version
 que le conviene a K-Means, porque ahora **todas las variables aportan de forma
 mas equilibrada** a la hora de comparar clientes.

```python
# %% Comparativa visual: histogramas SIN vs CON log1p
# Ponemos lado a lado la misma variable: izquierda cruda, derecha con log1p.
vars_demo = ['BALANCE', 'PURCHASES', 'CASH_ADVANCE', 'PAYMENTS']

fig, axes = plt.subplots(len(vars_demo), 2, figsize=(13, 16))
for i, col in enumerate(vars_demo):
    sns.histplot(datos_sin[col], bins=40, kde=True, ax=axes[i, 0], color='#C0504D')
    axes[i, 0].set_title(f'{col}  (SIN log1p)', fontsize=12, fontweight='bold')
    axes[i, 0].set_xlabel('')

    sns.histplot(datos_con[col], bins=40, kde=True, ax=axes[i, 1], color=AZUL)
    axes[i, 1].set_title(f'{col}  (CON log1p)', fontsize=12, fontweight='bold')
    axes[i, 1].set_xlabel('')

plt.suptitle('Efecto de log1p sobre la distribucion de las variables de dinero',
             fontsize=15, fontweight='bold', y=1.00)
plt.tight_layout()
plt.show()
```

```python
# %% Estandarizado (StandardScaler) sobre la version con log1p
# Despues de log1p seguimos necesitando poner todas las variables en la misma
# escala (media 0, desviacion 1) para que ninguna pese mas que otra por sus
# unidades. Esta es la matriz que usaremos de aqui en adelante.
escalador = StandardScaler()
X = escalador.fit_transform(datos_con)          # matriz final para PCA y K-Means
columnas = datos_con.columns

print('Matriz lista para clustering:', X.shape)
```

# 3.2 PCA: entender las variables antes de agrupar
 Tenemos 17 variables. Es imposible dibujar 17 dimensiones, y ademas muchas de
 ellas cuentan casi lo mismo (lo vimos en el heatmap, compras y frecuencia de
 compras iban de la mano). El **PCA (Analisis de Componentes Principales)**
 resume esas 17 variables en unos pocos "ejes nuevos" llamados componentes, que
 combinan las variables originales y guardan la mayor cantidad de informacion
 posible.

 Lo usamos con dos objetivos:
 1. **Entender** que variables se mueven juntas (con el circulo de correlacion).
 2. **Poder dibujar** a los clientes en un plano de 2 dimensiones para ver los
    grupos que encontrara K-Means.

```python
# %% Ajuste del PCA y varianza explicada (grafico de codo de la varianza)
pca = PCA()
componentes = pca.fit_transform(X)

var_exp = pca.explained_variance_ratio_ * 100      # % que aporta cada componente
var_acum = np.cumsum(var_exp)                       # % acumulado

fig, ax = plt.subplots(1, 2, figsize=(14, 5))

# Izquierda: cuanto aporta cada componente
ax[0].bar(range(1, len(var_exp) + 1), var_exp, color=AZUL, alpha=0.85)
ax[0].set_title('Varianza explicada por cada componente', fontweight='bold')
ax[0].set_xlabel('Componente')
ax[0].set_ylabel('% de informacion')

# Derecha: varianza acumulada
ax[1].plot(range(1, len(var_acum) + 1), var_acum, marker='o', color='#C0504D')
ax[1].axhline(80, color='gray', linestyle='--', label='80% de la informacion')
ax[1].set_title('Varianza acumulada', fontweight='bold')
ax[1].set_xlabel('Numero de componentes')
ax[1].set_ylabel('% acumulado')
ax[1].legend()

plt.tight_layout()
plt.show()

print('PC1:', round(var_exp[0], 1), '%  | PC2:', round(var_exp[1], 1),
      '%  | PC1+PC2:', round(var_acum[1], 1), '%')
print('Componentes para llegar al 80%:', int(np.argmax(var_acum >= 80) + 1))
```

### Lectura del Círculo de Correlación

El gráfico muestra la relación entre las variables y su proyección en las dos primeras componentes principales (**PC1** y **PC2**), las cuales explican en conjunto el **51.8% de la varianza** (29.9% + 21.9%).

#### 1. Grupos de variables identificados

* **Bloque de Compras (Eje Horizontal / Derecha):**
  Variables como `PURCHASES`, `PURCHASES_FREQUENCY`, `ONEOFF_PURCHASES`, `INSTALLMENTS_PURCHASES` y `PURCHASES_TRX` se extienden juntas hacia la derecha. Esto indica que **PC1 mide el volumen y frecuencia de compras**: a mayor valor a la derecha, más utiliza la tarjeta el cliente para comprar.

* **Bloque de Avances en Efectivo (Cuadrante Superior Izquierdo):**
  `CASH_ADVANCE`, `CASH_ADVANCE_FREQUENCY` y `CASH_ADVANCE_TRX` apuntan exactamente hacia la misma dirección. Esto confirma que el retiro de efectivo es un hábito recurrente y consistente entre ciertos usuarios.

* **Bloque de Deuda y Capacidad Crediticia (Eje Vertical / Arriba):**
  `BALANCE`, `MINIMUM_PAYMENTS`, `BALANCE_FREQUENCY`, `PAYMENTS` y `CREDIT_LIMIT` se orientan hacia la parte superior. Por lo tanto, **PC2 separa el nivel de saldo arrastrado y el límite crediticio**.

---

#### 2. Relaciones clave entre variables

* **Comportamiento opuesto:**
  `PRC_FULL_PAYMENT` (porcentaje de pago total) apunta hacia el cuadrante inferior derecho, en dirección contraria a `BALANCE` y `CASH_ADVANCE`. Esto refleja que los clientes que pagan la totalidad de su deuda no suelen arrastrar saldos ni solicitar avances en efectivo.

* **Independencia de hábitos:**
  El bloque de compras y el bloque de avances en efectivo forman un ángulo cercano a los 90° (perpendiculares), lo que demuestra que **comprar con la tarjeta y retirar efectivo son comportamientos independientes**.

```python
# %% Circulo de correlacion (que variable jala hacia donde)
# Correlacion de cada variable original con PC1 y PC2:
#   corr = vector propio * raiz(valor propio).  Como los datos estan
#   estandarizados, esto da las coordenadas dentro de un circulo de radio 1.
loadings = pca.components_.T * np.sqrt(pca.explained_variance_)

fig, ax = plt.subplots(figsize=(9, 9))

# Circulo de radio 1
circulo = plt.Circle((0, 0), 1, color='gray', fill=False, linestyle='--')
ax.add_artist(circulo)
ax.axhline(0, color='gray', lw=0.8)
ax.axvline(0, color='gray', lw=0.8)

for i, var in enumerate(columnas):
    x_i, y_i = loadings[i, 0], loadings[i, 1]
    ax.arrow(0, 0, x_i, y_i, head_width=0.03, head_length=0.03,
             color=AZUL, alpha=0.8, length_includes_head=True)
    ax.text(x_i * 1.12, y_i * 1.12, var, fontsize=9, ha='center', va='center')

ax.set_xlim(-1.15, 1.15)
ax.set_ylim(-1.15, 1.15)
ax.set_xlabel(f'PC1 ({var_exp[0]:.1f}%)', fontweight='bold')
ax.set_ylabel(f'PC2 ({var_exp[1]:.1f}%)', fontweight='bold')
ax.set_title('Circulo de correlacion de las variables', fontweight='bold')
ax.set_aspect('equal')
plt.tight_layout()
plt.show()
```

### 3.3 Codo de Jambu (metodo del codo) para elegir cuantos grupos
Lectura

En la curva de la inercia se ve como la bajada fuerte se suaviza alrededor de
**k = 3 o k = 4**: ahi esta el "codo". A partir de ese punto, sumar mas grupos
casi no mejora que tan compactos quedan. La silueta nos ayuda a desempatar y
confirmar cual de esos valores deja los grupos mejor separados.

> Elegimos el `k` del codo mirando ambos graficos. En la siguiente celda
> fijamos ese valor en la variable `K_ELEGIDO`.

```python
# %% Calculo del codo y de la silueta para k = 2..10
rango_k = range(2, 11)
inercias = []
siluetas = []

for k in rango_k:
    km = KMeans(n_clusters=k, random_state=42, n_init=10)
    etiquetas_k = km.fit_predict(X)
    inercias.append(km.inertia_)
    siluetas.append(silhouette_score(X, etiquetas_k))

fig, ax = plt.subplots(1, 2, figsize=(14, 5))

ax[0].plot(list(rango_k), inercias, marker='o', color=AZUL)
ax[0].set_title('Codo de Jambu (inercia)', fontweight='bold')
ax[0].set_xlabel('Numero de grupos (k)')
ax[0].set_ylabel('Inercia (que tan apretados)')

ax[1].plot(list(rango_k), siluetas, marker='o', color='#C0504D')
ax[1].set_title('Coeficiente de silueta', fontweight='bold')
ax[1].set_xlabel('Numero de grupos (k)')
ax[1].set_ylabel('Silueta (mas alto = mejor)')

plt.tight_layout()
plt.show()

for k, inc, sil in zip(rango_k, inercias, siluetas):
    print(f'k={k}:  inercia={inc:10.0f}   silueta={sil:.3f}')

```

# 3.4 K-Means final con el k elegido

```python
K_ELEGIDO = 4

kmeans = KMeans(n_clusters=K_ELEGIDO, random_state=42, n_init=10)
etiquetas = kmeans.fit_predict(X)

# Guardamos la etiqueta de grupo junto a los datos ORIGINALES (sin transformar)
# para poder interpretar cada grupo en sus unidades reales (dolares, %, etc.).
datos_grupos = datos.copy()
datos_grupos['CLUSTER'] = etiquetas

print('Clientes por grupo:')
print(datos_grupos['CLUSTER'].value_counts().sort_index())
print('\nSilueta final:', round(silhouette_score(X, etiquetas), 3))
```

### Lectura del Gráfico de Clústeres (K-Means, k=4)

El gráfico nos muestra cómo K-Means dividió a los clientes en 4 grupos, usando los dos ejes principales del PCA (**PC1** para el nivel de compras y **PC2** para el estilo financiero) para ubicarlos en el mapa.
#### 1. ¿Cómo se distribuye cada grupo?

* **Grupo 0 (Azul - Arriba a la derecha):**
  Son los clientes más activos. Compran mucho, usan la tarjeta seguido y manejan los límites de crédito más altos.

* **Grupo 1 (Naranja - A la izquierda):**
  Casi no usan la tarjeta para comprar, pero se concentran en la zona de retiro de efectivo y saldos pendientes. Usan el plástico más como un medio de financiamiento o emergencia que para compras cotidianas.

* **Grupo 2 (Verde - Centro abajo a la derecha):**
  Son compradores frecuentes y moderados, pero a diferencia del Grupo 0, mantienen sus deudas bajo control y no recurren al efectivo.

* **Grupo 3 (Rojo - En la parte más baja):**
  Están alineados con la zona de pago completo. Son los clientes "ordenados" que suelen liquidar el total de su cuenta mes a mes y evitan arrastrar saldos o pedir avances.


```python

# %% Visualizacion de los grupos en el plano PC1-PC2
plt.figure(figsize=(10, 7))
sns.scatterplot(x=componentes[:, 0], y=componentes[:, 1],
                hue=etiquetas, palette='tab10', s=18, alpha=0.6)
plt.xlabel(f'PC1 ({var_exp[0]:.1f}%)', fontweight='bold')
plt.ylabel(f'PC2 ({var_exp[1]:.1f}%)', fontweight='bold')
plt.title(f'Clientes agrupados por K-Means (k={K_ELEGIDO})', fontweight='bold')
plt.legend(title='Grupo')
plt.tight_layout()
plt.show()
```

```python
# %% Perfil de cada grupo (promedios en unidades reales)
# Comparamos los grupos en las variables mas faciles de interpretar.
perfil_vars = ['BALANCE', 'PURCHASES', 'CASH_ADVANCE', 'CREDIT_LIMIT',
               'PAYMENTS', 'PRC_FULL_PAYMENT', 'PURCHASES_FREQUENCY', 'TENURE']

perfil = datos_grupos.groupby('CLUSTER')[perfil_vars].mean().round(1)
perfil['N_CLIENTES'] = datos_grupos['CLUSTER'].value_counts().sort_index()
print(perfil.T)
```
