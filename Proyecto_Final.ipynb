## Análisis de tipos de variables — Dataset CC_GENERAL

El dataset contiene **18 columnas**, correspondientes al comportamiento de clientes de tarjetas de crédito.

### Tipos de datos

| Variable | Tipo | Descripción |
|---|---|---|
| `CUST_ID` | object | Identificador único del cliente (alfanumérico, no aporta valor analítico) |
| `BALANCE` | float64 | Saldo disponible en la cuenta para compras |
| `BALANCE_FREQUENCY` | float64 | Frecuencia de actualización del saldo (0 a 1) |
| `PURCHASES` | float64 | Monto total de compras realizadas |
| `ONEOFF_PURCHASES` | float64 | Monto de compras realizadas en un solo pago |
| `INSTALLMENTS_PURCHASES` | float64 | Monto de compras realizadas a cuotas |
| `CASH_ADVANCE` | float64 | Efectivo adelantado por el cliente con la tarjeta |
| `PURCHASES_FREQUENCY` | float64 | Frecuencia de compras (0 a 1) |
| `ONEOFF_PURCHASES_FREQUENCY` | float64 | Frecuencia de compras en un solo pago (0 a 1) |
| `PURCHASES_INSTALLMENTS_FREQUENCY` | float64 | Frecuencia de compras a cuotas (0 a 1) |
| `CASH_ADVANCE_FREQUENCY` | float64 | Frecuencia de adelantos de efectivo (0 a 1) |
| `CASH_ADVANCE_TRX` | int64 | Número de transacciones de adelanto de efectivo |
| `PURCHASES_TRX` | int64 | Número de transacciones de compra |
| `CREDIT_LIMIT` | float64 | Límite de crédito asignado al cliente |
| `PAYMENTS` | float64 | Monto de pagos realizados por el cliente |
| `MINIMUM_PAYMENTS` | float64 | Monto mínimo pagado por el cliente |
| `PRC_FULL_PAYMENT` | float64 | Porcentaje de veces que se pagó el saldo completo (0 a 1) |
| `TENURE` | int64 | Meses de antigüedad del cliente usando la tarjeta |

```python
datos.info
```

## 1.5 Limpieza de nulos

```python
nulos = datos.isnull().sum()
porcentaje_nulos = (nulos / len(datos)) * 100
resumen_nulos = pd.DataFrame({"Nulos": nulos, "% Nulos": porcentaje_nulos})
print(resumen_nulos[resumen_nulos["Nulos"] > 0])
```

## 1.6 Tratamiento de valores nulos y eliminacion de la columna "CUST_ID"

```python
datos = datos.drop(columns=["CUST_ID"])
```

```python

if datos["CREDIT_LIMIT"].isnull().sum() > 0:
    datos["CREDIT_LIMIT"] = datos["CREDIT_LIMIT"].fillna(datos["CREDIT_LIMIT"].median())

if datos["MINIMUM_PAYMENTS"].isnull().sum() > 0:
    datos["MINIMUM_PAYMENTS"] = datos["MINIMUM_PAYMENTS"].fillna(datos["MINIMUM_PAYMENTS"].median())


```

```python
print(datos.isnull().sum().sum(), "nulos restantes")
print(datos.shape)
datos.head()
```

## 1.7 Dataset Limpio

```python
datos.to_csv("CC_GENERAL_clean.csv", index=False)
```

## 2. Análisis exploratorio de datos (EDA)


## 2.1 Carga del dataset limpio

```python
datos = pd.read_csv('CC_GENERAL_clean.csv')
print(datos.shape)
datos.head()
```

## 2.2 Estadísticas descriptivas

```python
resumen_estadistico = datos.describe().T
resumen_estadistico['rango'] = resumen_estadistico['max'] - resumen_estadistico['min']
resumen_estadistico
```

## 2.3 Distribución de variables (histogramas)

**Nota sobre la selección de variables:** el dataset tiene 17 variables ya con la limpieza realizada, se seleccionaron 8 variables representativas que cubren las dimensiones clave del comportamiento del cliente:

- **Gasto:** `BALANCE`, `PURCHASES`
- **Uso del crédito como adelanto de efectivo:** `CASH_ADVANCE`
- **Capacidad crediticia:** `CREDIT_LIMIT`
- **Comportamiento de pago:** `PAYMENTS`, `PRC_FULL_PAYMENT`
- **Frecuencia de uso:** `PURCHASES_FREQUENCY`
- **Antigüedad:** `TENURE`

El resto de las variables se mantienen en el dataset completo para las estadísticas descriptivas, el heatmap de correlaciones, el escalado y la clusterización; solo se excluyen de los histogramas y boxplots para mantener la visualización legible.

```python
# Seleccionamos las variables mas relevantes para el analisis de comportamiento del cliente
variables_clave = ['BALANCE', 'PURCHASES', 'CASH_ADVANCE', 'CREDIT_LIMIT',
                    'PAYMENTS', 'PRC_FULL_PAYMENT', 'PURCHASES_FREQUENCY', 'TENURE']

fig, axes = plt.subplots(4, 2, figsize=(14, 20))
axes = axes.flatten()

for i, col in enumerate(variables_clave):
    sns.histplot(datos[col], bins=30, kde=True, ax=axes[i], color='#2E6DA4')
    axes[i].set_title(col, fontsize=13, fontweight='bold')
    axes[i].set_xlabel('')

plt.subplots_adjust(hspace=0.4, wspace=0.25)
plt.show()
```

## 2.4 Detección de valores atípicos (boxplots)

```python
# Boxplots de las mismas variables clave
fig, axes = plt.subplots(4, 2, figsize=(14, 20))
axes = axes.flatten()

for i, col in enumerate(variables_clave):
    sns.boxplot(y=datos[col], ax=axes[i], color='#2E6DA4')
    axes[i].set_title(col, fontsize=13, fontweight='bold')
    axes[i].set_ylabel('')

plt.subplots_adjust(hspace=0.4, wspace=0.25)
plt.show()
```

## 2.5 Matriz de correlaciones (heatmap)

```python
plt.figure(figsize=(20, 16))
matriz_corr = datos.corr()
sns.heatmap(matriz_corr, annot=True, fmt='.2f', cmap='coolwarm', linewidths=0.7,
            square=True, annot_kws={'size': 9}, cbar_kws={'shrink': 0.8})
plt.xticks(rotation=45, ha='right', fontsize=10)
plt.yticks(fontsize=10)
plt.title('Matriz de correlaciones entre variables', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

#### Aquí la realizamos con solamente las variables clave elegidas

```python
plt.figure(figsize=(20, 16))
matriz_corr = datos[variables_clave].corr()
sns.heatmap(matriz_corr, annot=True, fmt='.2f', cmap='coolwarm', linewidths=0.7,
            square=True, annot_kws={'size': 9}, cbar_kws={'shrink': 0.8})
plt.xticks(rotation=45, ha='right', fontsize=10)
plt.yticks(fontsize=10)
plt.title('Matriz de correlaciones entre variables clave', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

## 2.6 Escalado de datos

```python
escalador = StandardScaler()
datos_escalados = escalador.fit_transform(datos)
datos_escalados = pd.DataFrame(datos_escalados, columns=datos.columns)
datos_escalados.head()
```

```python
# Guardamos el dataset escalado
datos_escalados.to_csv('CC_GENERAL_scaled.csv', index=False)
print('Dataset escalado guardado como CC_GENERAL_scaled.csv')
print(datos_escalados.shape)
```

## 2.7 Comparación antes / después del escalado

```python
comparacion = pd.DataFrame({
    'media_original': datos.mean(),
    'std_original': datos.std(),
    'media_escalada': datos_escalados.mean().round(2),
    'std_escalada': datos_escalados.std().round(2)
})

fig, axes = plt.subplots(1, 2, figsize=(18, 7))

comparacion['media_original'].plot(kind='barh', ax=axes[0], color='#2E6DA4')
axes[0].set_title('Media de cada variable — ANTES del escalado', fontsize=12, fontweight='bold')
axes[0].set_xlabel('Media (escalas muy distintas entre variables)')

comparacion['media_escalada'].plot(kind='barh', ax=axes[1], color='#2E6DA4')
axes[1].set_title('Media de cada variable — DESPUÉS del escalado', fontsize=12, fontweight='bold')
axes[1].set_xlabel('Media (todas centradas en 0)')
axes[1].set_xlim(-1, 1)

plt.tight_layout()
plt.show()

comparacion
```
